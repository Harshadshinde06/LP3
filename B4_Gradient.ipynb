```python
import numpy as np
import matplotlib.pyplot as plt

# Function
def function(x):
    return (x + 3)**2

# Derivative
def gradient(x):
    return 2 * (x + 3)

print("Gradient Descent Algorithm")
print("Function: y = (x + 3)^2")
print("Library Imported Successfully!")
```

```python
x = 2
learning_rate = 0.1
iterations = 20

print("Starting Point:", x)
print("Learning Rate:", learning_rate)
print("Number of Iterations:", iterations)
```

```python
x_values = []
y_values = []

print("=" * 65)
print("             GRADIENT DESCENT ALGORITHM")
print("=" * 65)

print("\nIteration\tX\t\tY\t\tGradient")
print("-" * 65)

for i in range(iterations):

    y = function(x)
    grad = gradient(x)

    print(f"{i+1}\t\t{x:.6f}\t{y:.6f}\t{grad:.6f}")

    x_values.append(x)
    y_values.append(y)

    # Update x using Gradient Descent
    x = x - learning_rate * grad

print("-" * 65)

print("\nFinal Result")
print("Local Minimum x =", round(x, 6))
print("Minimum Value y =", round(function(x), 6))
print("=" * 65)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, iterations + 1),
    y_values,
    marker="o"
)

plt.xlabel("Iterations")
plt.ylabel("Function Value")
plt.title("Gradient Descent Convergence")
plt.grid(True)

plt.show()
```

```python
x_range = np.linspace(-6, 4, 100)
y_range = function(x_range)

plt.figure(figsize=(8, 5))

plt.plot(x_range, y_range, label="y = (x + 3)^2")

plt.scatter(
    x,
    function(x),
    color="red",
    label="Minimum Point"
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent - Local Minimum")
plt.legend()
plt.grid(True)

plt.show()
```

```python
print("Gradient Descent Algorithm Implemented Successfully!")
print("The function is y = (x + 3)^2")
print("Starting point is x = 2")
print("The algorithm converges towards x = -3")
print("Minimum value of the function is approximately 0")
```
